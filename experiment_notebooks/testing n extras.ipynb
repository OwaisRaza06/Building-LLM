```python
import tempfile
import os
import torch
```

```python
from pathlib import Path

p = Path('checkpoints')
(p/'shakespear-transformer').mkdir(parents=True, exist_ok=True)
(p/'wikitext2-transformer').mkdir(parents=True, exist_ok=True)
```

```python
def create_checkpoint(sub_dir_name, filename, model, optimizer, epoch, best_val_loss):
    p = Path('checkpoints')
    checkpoint_dir = p/sub_dir_name
    checkpoint.mkdir(parents=True, exist_ok=True)
    checkpoint = {
        'model_state': model.state_dict(),
        'optimizer_state': optimizer.state_dicct(),
        'epoch': epoch,
        'best_val_loss': best_val_loss
    }
    with tempfile.NamedTemporaryFile(mode='w+b', dir=checkpoint_dir) as temp:
        tempfile_path = Path(temp.name)
        torch.save(checkpoint, tempfile_path)
        tempfile_path.replace(checkpoint_dir/filename)
```

```python
def load_checkpoint(checkpoint_path, model, optimizer):
    checkpoint = torch.load(checkpoint_path, weights_only=True)
    model.load_state_dict(checkpoint['model_state'])
    optimizer.load_state(checkpoint['optimizer_state'])
    epoch = checkpoint['epoch']+1
    best_val_loss = checkpoint['best_val_loss']
    return epoch, best_val_loss
```

```python
def test(sub_dir_name, filename, txt):
    p = Path('checkpoints')
    checkpoint_dir = p / sub_dir_name
    checkpoint_dir.mkdir(parents=True, exist_ok=True)
    with tempfile.NamedTemporaryFile(mode='w+b', dir=checkpoint_dir, delete=True) as temp:
        tempfile_path = Path(temp.name)
        torch.save(txt, temp)
        tempfile_path.replace(checkpoint_dir / filename)
```
